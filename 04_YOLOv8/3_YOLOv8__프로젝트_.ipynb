Filtering Classes + Labels

--> Keeps only the classes the scenario needs and adds text labels.

```python
img_copy = img_np.copy()


for det in results[0].boxes.data:
    x1, y1, x2, y2, conf, cls = det.tolist()
    if int(cls) == 0:
        x1, y1, x2, y2 = map(int, [x1, y1, x2, y2])
        label = f"{class_names[int(cls)]} {conf:.2f}"
        cv2.rectangle(img_copy, (x1, y1), (x2, y2), (255, 0, 0), 2)
        cv2.putText(img_copy, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX,
                    0.9, (255, 0, 0), 2)


plt.figure(figsize=(10, 8))
plt.imshow(img_copy)
plt.title("Only Class 0 (person) Bounding Boxes with Labels")
plt.axis("off")
plt.show()
```

Cuts detected objects into separate image files, for example to pass them to another model.

```python
import os
bus_class_id = [k for k, v in class_names.items() if v == 'bus'][0]
os.makedirs("bus_crops", exist_ok=True)

count = 0
for det in results[0].boxes.data:
    x1, y1, x2, y2, conf, cls = det.tolist()
    if int(cls) == bus_class_id:
        x1, y1, x2, y2 = map(int, [x1, y1, x2, y2])
        cropped = img_np[y1:y2, x1:x2]
        save_path = f"bus_crops/bus_crop_{count}.jpg"
        cv2.imwrite(save_path, cv2.cvtColor(cropped, cv2.COLOR_RGB2BGR))
        print(f"버스 {count}번 잘라서 저장: {save_path}")
        count += 1
```

```python
import os
import cv2
import matplotlib.pyplot as plt

crop_folder = "bus_crops"

image_files = sorted([f for f in os.listdir(crop_folder) if f.endswith(".jpg")])

num_images = len(image_files)
plt.figure(figsize=(5 * num_images, 5))

for i, filename in enumerate(image_files):
    img_path = os.path.join(crop_folder, filename)
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # OpenCV는 BGR로 불러오기 때문에 RGB로 변환

    plt.subplot(1, num_images, i + 1)
    plt.imshow(img)
    plt.title(filename)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

## A: Model Size

```python
print(dataset.location)
!ls {dataset.location}
```

```python
DATASET_DIR = dataset.location          # instead of "/content/stamp-10"
DATA = f"{DATASET_DIR}/data.yaml"
```

```python
sizes = ["n", "s", "m"]
size_rows = []

for s in sizes:
  model = YOLO(f"yolov8{s}.pt")
  start = time.time()
  model.train(data=DATA, epochs=20, imgsz=640, batch=16, device=0, project=f"{SAVE_DIR}/runs", name=f"size_{s}", exist_ok=True)
  train_min = (time.time() - start) / 60

  best_path = f"{SAVE_DIR}/runs/size_{s}/weights/best.pt"
  best = YOLO(best_path)
  m = best.val(data=DATA, split="val", verbose=False)

  size_rows.append({
      "model": f"yolov8{s}",
      "params (M)": sum(p.numel() for p in best.model.parameters()) / 1e6,
      "file (MB)": os.path.getsize(best_path) / 1e6,
      "train time (min)": train_min,
      "inference (ms/img)": m.speed["inference"],
      "mAP50": m.box.map50,
      "mAP50-95": m.box.map,
      "precision": m.box.mp,
      "recall": m.box.mr,
  })
  pd.DataFrame(size_rows).round(3).to_csv(f"{SAVE_DIR}/A_model_size.csv", index=False)

df_size = pd.DataFrame(size_rows).round(3)
df_size
```

## B: Confidence Threshold

```python
best = YOLO(f"{SAVE_DIR}/runs/size_n/weights/best.pt")
test_img_dir = f"{DATASET_DIR}/test/images"
test_lbl_dir = f"{DATASET_DIR}/test/labels"

preds = best.predict(test_img_dir, conf=0.01, verbose=False)

def load_gt(img_path, w, h):
  lbl = Path(test_lbl_dir) / (Path(img_path).stem + ".txt")
  boxes = []
  if lbl.exists():
    for line in lbl.read_text().strip().splitlines():
      c, cx, cy, bw, bh = map(float, line.split()[:5])
      boxes.append([c, (cx-bw/2)*w, (cy-bh/2)*h, (cx+bw/2)*w, (cy+bh/2)*h])
  return np.array(boxes).reshape(-1,5)

def iou(a, b):
    x1, y1 = max(a[0], b[0]), max(a[1], b[1])
    x2, y2 = min(a[2], b[2]), min(a[3], b[3])
    inter = max(0, x2 - x1) * max(0, y2 - y1)
    union = (a[2]-a[0])*(a[3]-a[1]) + (b[2]-b[0])*(b[3]-b[1]) - inter
    return inter / union if union > 0 else 0

def evaluate(conf_thr, iou_thr=0.5):
  tp = fp = fn = 0
  for r in preds:
    h,w = r.orig_shape
    gt = load_gt(r.path, w, h)
    d = r.boxes.data.cpu().numpy()
    d = d[d[:, 4] >= conf_thr]
    matched = set()
    for x1, y1, x2, y2, conf, cls in d:
            best_iou, best_j = 0, -1
            for j, g in enumerate(gt):
              if j in matched or int(g[0]) != int(cls):
                continue
              v = iou([x1, y1, x2, y2], g[1:])
              if v > best_iou:
                best_iou, best_j = v, j
            if best_iou >= iou_thr:
              tp += 1
              matched.add(best_j)
            else:
              fp += 1
    fn += len(gt) - len(matched)
  p = tp / (tp + fp) if tp + fp else 0
  r_ = tp / (tp + fn) if tp + fn else 0
  f1 = 2 * p * r_ / (p + r_) if p + r_ else 0
  return {"conf": conf_thr, "TP": tp, "FP (false alarms)": fp, "FN (missed)": fn,
            "precision": p, "recall": r_, "F1": f1}

thresholds = [0.05, 0.1, 0.25, 0.5, 0.7, 0.9]
df_conf = pd.DataFrame([evaluate(t) for t in thresholds]).round(3)
df_conf.to_csv(f"{SAVE_DIR}/B_conf_thr.csv", index=False)
df_conf
```

```python
import matplotlib.pyplot as plt
plt.plot(df_conf["conf"], df_conf["precision"], "o-", label="precision")
plt.plot(df_conf["conf"], df_conf["recall"], "o-", label="recall")
plt.plot(df_conf["conf"], df_conf["F1"], "o-", label="F1")
plt.xlabel("confidence threshold"); plt.legend(); plt.grid(True); plt.show()
```
